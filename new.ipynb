```python
#X GATE
from qiskit import QuantumCircuit
from qiskit.providers.aer import QasmSimulator
qc=QuantumCircuit(1,1)
qc.x(0)
qc.measure(0,0)
simulator=QasmSimulator()
result=simulator.run(qc,shots=10).result()
print(qc)
print(result.get_counts())
```

```python
#REVERSE CNOT
from qiskit import QuantumCircuit
from qiskit.providers.aer import QasmSimulator
qc=QuantumCircuit(2,2)
qc.x(1)
qc.cx(1,0)
qc.measure(0,0)
qc.measure(1,1)
simulator=QasmSimulator()
result=simulator.run(qc,shots=10).result()
print(qc)
print(result.get_counts())
```

```python
#SWAP GATE
from qiskit import QuantumCircuit
from qiskit.providers.aer import QasmSimulator
qc=QuantumCircuit(2,2)
qc.x(1)
qc.swap(1,0)
qc.measure(0,0)
qc.measure(1,1)
simulator=QasmSimulator()
result=simulator.run(qc,shots=10).result()
print(qc)
print(result.get_counts())
```

```python
#TOFFOLI GATE[2 control qubits and one target qubit]
from qiskit import QuantumCircuit
from qiskit.providers.aer import QasmSimulator
qc=QuantumCircuit(3,3)
qc.x(0)
qc.x(1)
qc.ccx(0,1,2)
qc.measure(0,0)
qc.measure(1,1)
qc.measure(2,2)
simulator=QasmSimulator()
result=simulator.run(qc,shots=10).result()
print(qc)
print(result.get_counts())
```
